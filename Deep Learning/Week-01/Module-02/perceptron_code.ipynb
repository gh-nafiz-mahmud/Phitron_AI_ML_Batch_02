```python
# and gate dataset

x = [(0,0), (0,1), (1,0),(1,1)]
y = [0,0,0,1]

# initial weights and bias
w1, w2 = 0, 0
b = 0
lr = 0.1

# training loop
for epoch in range(10):
    for i in range(len(x)):
        x1, x2 = x[i]
        target = y[i]

        weighted_sum = w1*x1 + x2*x2 + b
        prediction = 1 if weighted_sum >= 0 else 1

        error = target - prediction

        # weight update 
        w1 = w1 + lr * error * x1
        w2 = w2 + lr * error * x2
        b = b + lr * error

print(w1, w2, b)
print(prediction)
print(error)
```
